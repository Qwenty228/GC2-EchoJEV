```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt

artifacts = sorted(p for p in Path("artifacts/a2").glob("*/") if (p / "COMPLETE").exists())
assert artifacts, "Train A2 first"
artifact = artifacts[-1]
print("Model:", artifact)
```

```python
weights = np.array(json.loads((artifact / "layer_weights.json").read_text()))
labels = ["Embedding"] + [f"L{i}" for i in range(1, len(weights))]
plt.figure(figsize=(10, 3))
plt.bar(labels, weights)
plt.title("A2 learned WavLM layer weights (global, not per-clip explanations)")
plt.ylabel("Softmax weight")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
history = json.loads((artifact / "training_curve.json").read_text())
plt.plot(history["train_loss"], label="Train")
plt.plot(history["dev_loss"], label="Dev")
plt.axvline(history["best_epoch"] - 1, linestyle="--", label="Best epoch")
plt.xlabel("Epoch (0-based)")
plt.ylabel("Weighted BCE")
plt.legend()
plt.show()
```
